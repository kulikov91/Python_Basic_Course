## Требования

**Предусловия**
- Python 3.11+
- Файлы `Farm_proj.py`, `Export.csv`, `reviews.csv`.

**Развертывание**
```bash
git clone <repository>
cd <repository>
python Farm_proj.py
```

Если Git не используется, распакуйте архив и запустите `Farm_proj.py`.


## Структура проекта

- Farm_proj.py
- Export.csv
- reviews.csv

Классы:
- Market
- MarketManager
- ReviewManager
- FarmMarketApp


## Безопасная эксплуатация

**Предусловия**
- Не изменять структуру CSV.
- Использовать UTF-8.

**Постусловия**
- Данные читаются корректно.
- Отзывы сохраняются без потерь.

Рекомендуется регулярно создавать резервную копию `reviews.csv`.


## Возможные ошибки

| Ошибка | Причина | Решение |
|---|---|---|
| FileNotFoundError | Нет CSV | Поместить файл рядом с программой |
| ValueError | Некорректный ввод | Ввести число |
| Рынок не найден | Нет данных | Проверить FMID |
