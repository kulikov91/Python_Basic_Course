## 1. Назначение программы

Приложение реализует игру «Морской бой» с графическим интерфейсом Tkinter.

Пользователь играет против компьютера. Слева отображается поле игрока, справа — поле компьютера. Игрок стреляет по клеткам поля компьютера мышью.

## 2. Запуск

Запуск в Windows:

```bat
run_windows.bat
```

Запуск в Linux:

```bash
chmod +x run_linux.sh
./run_linux.sh
```

Также можно запустить напрямую:

```bash
python battleship.py
```

## 3. Главное окно

В главном окне находятся:

- поле игрока;
- поле компьютера;
- строка состояния;
- кнопки управления;
- меню приложения.

![Главное окно](screenshots/main_window.png)

## 4. Правила игры

Корабли игрока показаны на левом поле. Корабли компьютера скрыты.

Чтобы сделать ход, нужно нажать на клетку правого поля. После хода игрока компьютер автоматически делает ответный ход.

Цвета клеток:

| Вид клетки | Значение |
|---|---|
| синий | корабль игрока |
| красный `X` | попадание |
| серый `•` | промах |
| голубой | неизвестная вода или скрытый корабль компьютера |

![Окно правил](screenshots/help_window.png)

## 5. Новая игра

Кнопка **Новая игра** очищает старую партию, заново создаёт поля и случайно расставляет корабли.

При начале новой игры используются текущие настройки из `config.json`.

## 6. Настройки

Окно настроек открывается через кнопку **Настройки** или меню `Игра → Настройки`.

Пользователь может изменить:

- имя игрока;
- размер поля;
- размер клетки;
- количество кораблей;
- сложность;
- размер шрифта.

После сохранения настройки записываются в `config.json`.

![Окно настроек](screenshots/settings_window.png)

## 7. Лучшие результаты

Окно **Лучшие результаты** показывает победные результаты, отсортированные по количеству ходов. Чем меньше ходов, тем выше результат.

Результаты сохраняются в `scores.json`.

![Лучшие результаты](screenshots/scores_window.png)

## 8. Завершение работы

Закрыть приложение можно через:

- кнопку закрытия окна;
- меню `Игра → Выход`.

Настройки и результаты сохраняются в JSON-файлах.

## 9. Возможные сообщения

| Сообщение | Причина |
|---|---|
| `В эту клетку уже стреляли` | пользователь повторно нажал на ту же клетку |
| `Вы победили` | все корабли компьютера уничтожены |
| `Вы проиграли` | компьютер уничтожил все корабли игрока |
| ошибка настроек | введены недопустимые значения размера поля, клетки или количества кораблей |
