## 2. Требования

Приложение использует стандартную библиотеку Python:

- `tkinter`;
- `json`;
- `random`;
- `os`.

Сторонние библиотеки не требуются.

Для Linux может понадобиться установка Tkinter отдельно, например:

```bash
sudo apt install python3-tk
```

## 3. Развёртывание в Windows

1. Установить Python 3.
2. При установке отметить пункт `Add Python to PATH`.
3. Распаковать архив проекта.
4. Запустить:

```bat
run_windows.bat
```

Если окно не открылось, запустить из командной строки:

```bat
python battleship.py
```

## 4. Развёртывание в Linux

1. Установить Python 3 и Tkinter.
2. Распаковать архив проекта.
3. Разрешить запуск скрипта:

```bash
chmod +x run_linux.sh
```

4. Запустить:

```bash
./run_linux.sh
```

## 5. Конфигурация JSON

Настройки хранятся в `config.json`.

Пример:

```json
{
    "player_name": "Player",
    "board_size": 10,
    "cell_size": 32,
    "ship_count": 10,
    "difficulty": "easy",
    "water_color": "#bde0fe",
    "hit_color": "#e63946",
    "miss_color": "#eeeeee",
    "ship_color": "#457b9d",
    "font_size": 10
}
```

Если файл отсутствует, программа создаёт настройки по умолчанию.

## 6. Результаты игроков

Результаты хранятся в `scores.json`.

Файл содержит список записей:

```json
[
    {
        "name": "Player",
        "result": "Победа",
        "moves": 25,
        "board_size": 10,
        "difficulty": "easy"
    }
]
```

Лучшие результаты определяются по минимальному количеству ходов среди победных партий.

## 7. Проверка работоспособности

Проверка синтаксиса:

```bash
python -m py_compile battleship.py
```

Запуск doctest:

```bash
python -m doctest -v battleship.py
```

Проверить вручную:

1. запуск главного окна;
2. открытие настроек;
3. сохранение настроек;
4. начало новой игры;
5. открытие лучших результатов;
6. открытие правил.

## 8. Резервное копирование

Перед заменой версии приложения рекомендуется сохранить:

- `config.json`;
- `scores.json`.

Пример Linux:

```bash
cp config.json config_backup.json
cp scores.json scores_backup.json
```

Пример Windows:

```bat
copy config.json config_backup.json
copy scores.json scores_backup.json
```

## 9. Безопасная и стабильная эксплуатация

Рекомендации:

- не удалять `config.json` и `scores.json` во время работы программы;
- не редактировать JSON-файлы при открытом приложении;
- хранить резервную копию результатов;
- после изменения кода запускать doctest;
- не использовать слишком большие размеры поля, чтобы интерфейс помещался на экран;
- запускать приложение из каталога проекта.

## 10. Диагностика проблем

| Проблема | Возможная причина | Решение |
|---|---|---|
| окно не открывается | не установлен Tkinter | установить `python3-tk` |
| настройки не сохраняются | нет прав записи | проверить права на папку проекта |
| ошибка JSON | файл повреждён | удалить файл, программа создаст настройки заново |
| результаты пропали | удалён `scores.json` | восстановить из резервной копии |
| команда `python` не найдена | Python не добавлен в PATH | переустановить Python или использовать полный путь |
