## 1. Назначение приложения

Программа предназначена для работы со справочником фермерских рынков США.

Пользователь может:

- просматривать рынки постранично;
- искать рынки по городу и штату;
- искать рынки по ZIP-коду;
- находить рынки в заданном радиусе;
- просматривать подробные сведения о рынке;
- добавлять и просматривать отзывы;
- видеть средний рейтинг рынка;
- удалять отзывы и рынки;
- сортировать рынки по названию, городу, штату, рейтингу и расстоянию.

Команды вводятся с клавиатуры без кавычек. Регистр команд не имеет значения.

## 2. Запуск программы

Для работы необходимо, чтобы в одной папке находились:

- `Farm_proj_OOP.py`;
- `Export.csv`;
- `reviews.csv`;
- `zip_util.py`;
- `zip_codes_states.csv`.

Запуск из терминала:

```bash
python Farm_proj_OOP.py
```

После запуска появляется приветственное сообщение и список команд.

![Запуск программы](screenshots/01_start.png)

## 3. Главное меню

Основные команды:

| Команда | Назначение |
|---|---|
| `show` | просмотр списка рынков |
| `search` | поиск по городу и штату |
| `zip` | поиск по ZIP-коду |
| `distance` | поиск в радиусе от ZIP-кода |
| `details` | подробная информация о рынке |
| `delete` | удаление рынка или отзыва |
| `review` | добавление отзыва |
| `sort` | сортировка рынков |
| `reviews` | просмотр отзывов |
| `help` | повторный вывод списка команд |
| `end` | завершение программы |

## 4. SHOW — просмотр списка рынков

Введите:

```text
show
```

Программа сообщает количество рынков, размер страницы и число доступных страниц. После ввода номера страницы выводятся:

- FMID;
- название рынка;
- город;
- штат;
- средний рейтинг.

В режиме просмотра доступны команды:

- `n` — следующая страница;
- `p` — предыдущая;
- номер страницы — перейти непосредственно на неё;
- `q` — выйти из просмотра.

![Просмотр списка рынков](screenshots/02_show.png)

## 5. SEARCH — поиск по городу и штату

Введите `search`, затем название города и штат.

Штат допускается вводить полным названием или двухбуквенным кодом. Например:

```text
Введите город: Danville
Введите штат: VT
```

Если рынки найдены, программа выводит их FMID, названия и рейтинг. После этого можно ввести FMID для просмотра подробностей либо нажать Enter.

![Поиск по городу и штату](screenshots/03_search.png)

## 6. ZIP — поиск по ZIP-коду

Введите `zip`, затем ZIP-код, например `05828`.

Программа выводит найденные рынки и их рейтинг. После поиска можно перейти к подробной информации по FMID.

![Поиск по ZIP-коду](screenshots/04_zip.png)

## 7. DISTANCE — поиск в заданном радиусе

Введите `distance`. Затем:

1. укажите исходный ZIP-код;
2. укажите радиус в милях.

Программа использует координаты ZIP-кода и рассчитывает расстояние до рынков. Результаты выводятся от ближайшего к более удалённому.

Отрицательный радиус не допускается. Если ZIP-код отсутствует в справочнике, программа сообщает об этом.

![Поиск по расстоянию](screenshots/05_distance.png)

## 8. DETAILS — подробная информация

Введите `details`, затем FMID рынка.

Программа показывает:

- название;
- город и штат;
- адрес;
- ZIP-код;
- сайт;
- доступные категории товаров;
- способы оплаты и программы поддержки;
- средний рейтинг;
- количество отзывов.

![Подробная информация](screenshots/06_details.png)

## 9. REVIEW — добавление отзыва

Введите `review`, затем:

1. FMID рынка;
2. имя и фамилию;
3. оценку от 1 до 5;
4. текст отзыва.

Текст может занимать несколько строк. Для завершения ввода текста необходимо ввести точку `.` в отдельной строке.

Оценка обязательна. Пустое имя не допускается.

![Добавление отзыва](screenshots/07_review.png)

## 10. REVIEWS — просмотр отзывов

Введите `reviews`, затем FMID рынка.

Если отзывы существуют, программа показывает средний рейтинг и каждый отзыв: автора, оценку и текст.

![Просмотр отзывов](screenshots/08_reviews.png)

## 11. DELETE — удаление

После команды `delete` программа предлагает:

```text
1 - Отзыв
2 - Рынок
0 - Отмена
```

### Удаление отзыва

Выберите `1`, введите FMID, затем номер отзыва. Перед удалением программа запрашивает подтверждение.

### Удаление рынка

Выберите `2`, введите FMID. Программа показывает сведения о рынке и количество связанных отзывов. После подтверждения рынок удаляется из `Export.csv`, а связанные отзывы — из `reviews.csv`.

**Внимание:** удаление рынка изменяет исходный файл данных. Перед массовыми изменениями рекомендуется создать резервную копию.

![Удаление](screenshots/10_delete.png)

## 12. SORT — сортировка

Введите `sort`. Доступны пять критериев:

1. название рынка;
2. город;
3. штат;
4. средний рейтинг;
5. расстояние от указанного ZIP-кода.

Направление:

- `a` — по возрастанию;
- `d` — по убыванию.

Для первых четырёх вариантов после сортировки указывается номер страницы. При сортировке по расстоянию программа запрашивает ZIP-код и показывает первые 20 результатов.

![Сортировка](screenshots/09_sort.png)

## 13. HELP

Команда:

```text
help
```

повторно выводит приветственное сообщение и полный список доступных команд.

## 14. END

Для штатного завершения программы используйте:

```text
end
```

После этого появляется сообщение:

```text
Программа завершена!
```

## 15. Сообщения об ошибках и рекомендации

Программа проверяет основные ошибки пользовательского ввода.

| Ситуация | Реакция программы |
|---|---|
| неверный номер страницы | `Такой страницы нет!` |
| вместо номера введён текст | просьба ввести значение цифрами |
| неизвестный FMID | `Рынок ... не найден!` |
| неизвестный ZIP | сообщение об отсутствии ZIP-кода |
| отрицательный радиус | радиус не принимается |
| рейтинг вне 1–5 | программа повторно запрашивает оценку |
| неизвестная команда | предлагается использовать `help` |

Для сохранности данных рекомендуется завершать работу командой `end` и не редактировать CSV-файлы во время работы программы.
